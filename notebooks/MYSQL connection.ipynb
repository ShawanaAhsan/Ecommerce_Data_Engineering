# Connect Python to MySQL

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "mysql+pymysql://root:shawana123@localhost/ecommerce_dw")
```

```python
engine.connect()
```

# Prepare dimension tables

```python
df = pd.read_csv("cleaned_data.csv")
```

```python
customers = df[["customer_id"]].drop_duplicates()
```

```python
customers.to_sql(
    "dim_customer",
    engine,
    if_exists="append",
    index=False
)
```

```python
products = df[["product_category"]].drop_duplicates()
```

```python
products.to_sql(
    "dim_product",
    engine,
    if_exists="append",
    index=False
)
```

```python
payments = df[["payment_method"]].drop_duplicates()
```

```python
payments.to_sql(
    "dim_payment",
    engine,
    if_exists="append",
    index=False
)
```

# Get MySQL IDs

```python
product_lookup = pd.read_sql(
    "SELECT product_id, product_category FROM dim_product",
    engine
)
```

```python
payment_lookup = pd.read_sql(
    "SELECT payment_id, payment_method FROM dim_payment",
    engine
)
```

```python
sales = df.merge(
    product_lookup,
    on="product_category",
    how="left"
)
```

```python
sales = sales.merge(
    payment_lookup,
    on="payment_method",
    how="left"
)
```

```python
fact_sales = sales[
    [
        "order_id",
        "order_date",
        "customer_id",
        "product_id",
        "region",
        "quantity",
        "unit_price",
        "discount",
        "payment_id",
        "delivery_days",
        "customer_rating",
        "revenue",
        "gross_sales",
        "discount_amount",
        "calculated_revenue"
    ]
]
```

```python
fact_sales.to_sql(
    "fact_sales",
    engine,
    if_exists="append",
    index=False
)
```
